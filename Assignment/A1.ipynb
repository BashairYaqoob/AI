```python
# Q6 a (i)
#Depth-First Search with a depth limit
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': [],
    'F': ['I'],
    'G': [],
    'H': [],
    'I': []
}

def dls(graph, start, goal, depth_limit):
    visited = []
    def dfs(node, depth):
        if depth > depth_limit:
            return None
        visited.append(node)
        if node == goal:
            print(f"Goal found with DLS. Path: {visited}")

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                path = dfs(neighbor, depth + 1)
                if path:
                    return path
        visited.pop()
        return None

    return dfs(start, 0)

dls(graph, 'A', 'I', 3)
```

```python
# Q6 a (ii)
# Iterative Deepening Search
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': [],
    'F': ['I'],
    'G': [],
    'H': [],
    'I': []
}

def dls(graph, node, goal, depth_limit, visited):
    if depth_limit < 0:
        return None

    visited.append(node)

    if node == goal:
        return visited.copy()

    for neighbor in graph.get(node, []):
        if neighbor not in visited:
            path = dls(graph, neighbor, goal, depth_limit - 1, visited)
            if path:
                return path

    visited.pop()
    return None

def ids(graph, start, goal, max_depth):
    for depth in range(max_depth + 1):
        print(f"\nDepth Limit = {depth}")
        visited = []
        path = dls(graph, start, goal, depth, visited)

        if path:
            print("Goal Found!")
            print("Path:", path)
            return path

    print("Goal not found within depth limit")
    return None

start_node = 'A'
goal_node = 'I'
max_depth = 4

print("Iterative Deepening Search:")
ids(graph, start_node, goal_node, max_depth)
```

```python
# Q6 b
def find_positions(maze):
    start = None
    goal = None

    for i in range(len(maze)):
        for j in range(len(maze[0])):
            if maze[i][j] == 'S':
                start = (i, j)
            elif maze[i][j] == 'G':
                goal = (i, j)

    return start, goal

def depth_limited_dfs(maze, start, goal, depth_limit):

    rows = len(maze)
    cols = len(maze[0])

    def dfs(position, depth, path, visited):

        if depth > depth_limit:
            return None

        x, y = position
        path.append(position)
        visited.append(position)

        if position == goal:
            return path.copy()

        moves = [(-1,0), (1,0), (0,-1), (0,1)]

        for dx, dy in moves:
            nx, ny = x + dx, y + dy
            if (0 <= nx < rows and 0 <= ny < cols and maze[nx][ny] != 1 and (nx, ny) not in visited):

                result = dfs((nx, ny), depth + 1, path, visited)
                if result:
                    return result

        path.pop()
        return None

    return dfs(start, 0, [], [])

def iterative_deepening_search(maze, start, goal, max_depth):

    for depth in range(max_depth + 1):
        print(f"\nDepth Limit = {depth}")

        result = depth_limited_dfs(maze, start, goal, depth)

        if result:
            print("Goal Found!")
            return result

    print("Goal not found within depth limit")
    return None

maze = [
['S', 0, 0, 1, 0],
[1, 0, 1, 0, 0],
[0, 0, 0, 0, 'G'],
[1, 1, 0, 1, 1]
]

start, goal = find_positions(maze)

dls_path = depth_limited_dfs(maze, start, goal, depth_limit=10)
print("\nDLS Path:", dls_path)
ids_path = iterative_deepening_search(maze, start, goal, max_depth=10)
print("IDS Path:", ids_path)
```
