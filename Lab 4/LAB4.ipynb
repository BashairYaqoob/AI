```python
#Task 1a: Depth-Limited Search (DLS) as Goal-Based Agent
tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': [],
    'F': ['I'],
    'G': [],
    'H': [],
    'I': []
}

class GoalBasedAgent:
    def __init__(self, goal, depth_limit):
        self.goal = goal
        self.depth_limit = depth_limit

    def formulate_goal(self, percept):
        if percept == self.goal:
            return 'goal reached'
        else:
            return 'searching'

    def dls(self, graph, node, goal, depth, visited):
        print("Visited:", node)

        if node == goal:
            print("Goal found!")
            return True

        if depth == 0:
            return False

        visited.append(node)

        for child in graph[node]:
            if child not in visited:
                found = self.dls(graph, child, goal, depth - 1, visited)
                if found:
                    return True

        return False

    def act(self, percept, graph):
        action = self.formulate_goal(percept)

        if action == 'goal reached':
            return f'{self.goal} is found'
        else:
            found = self.dls(graph, percept, self.goal, self.depth_limit, [])
            if not found:
                return "Goal not found within depth limit"


class Environment:
    def __init__(self, graph):
        self.graph = graph

    def get_percept(self, node):
        return node


def run_agent(agent, environment, start_node):
    percept = environment.get_percept(start_node)
    action = agent.act(percept, environment.graph)
    print(action)


start_node = 'A'
goal_node = 'E'
depth_limit = 2

g = GoalBasedAgent(goal_node, depth_limit)
e = Environment(tree)
run_agent(g, e, start_node)
```

```python
#Task 1b: Uniform Cost Search (UCS) as Utility-Based Agent

import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 1), ('G', 3)],
    'D': [('H', 1)],
    'E': [],
    'F': [('I', 2)],
    'G': [],
    'H': [],
    'I': []
}

class UtilityBasedAgent:
    def __init__(self, goal):
        self.goal = goal

    def formulate_utility(self, cost):
        return cost  # Lower cost = better utility

    def ucs(self, graph, start, goal):
        visited = set()
        pq = []
        heapq.heappush(pq, (0, start))

        while pq:
            cost, node = heapq.heappop(pq)

            if node in visited:
                continue

            print("Visited:", node, "Cost:", cost)
            visited.add(node)

            if node == goal:
                print("Goal found with minimum cost!")
                return cost

            for neighbor, weight in graph[node]:
                if neighbor not in visited:
                    heapq.heappush(pq, (cost + weight, neighbor))

        return None

    def act(self, percept, graph):
        cost = self.ucs(graph, percept, self.goal)
        if cost is not None:
            return f'{self.goal} found with minimum cost {cost}'
        else:
            return "Goal not reachable"


class Environment:
    def __init__(self, graph):
        self.graph = graph

    def get_percept(self, node):
        return node


def run_agent(agent, environment, start_node):
    percept = environment.get_percept(start_node)
    action = agent.act(percept, environment.graph)
    print(action)


start_node = 'A'
goal_node = 'H'

u = UtilityBasedAgent(goal_node)
e = Environment(graph)
run_agent(u, e, start_node)
```

```python
#Task 2
import itertools

# Cities from image
cities = [1, 2, 3, 4]

# Distance matrix
dist = {
    (1, 2): 10, (2, 1): 10,
    (1, 3): 15, (3, 1): 15,
    (1, 4): 20, (4, 1): 20,
    (2, 4): 25, (4, 2): 25,
    (4, 3): 30, (3, 4): 30,
    (2, 3): 35, (3, 2): 35,
}

def tsp(start):
    min_path = None
    min_cost = float('inf')

    other_cities = [c for c in cities if c != start]

    for perm in itertools.permutations(other_cities):
        route = [start] + list(perm) + [start]
        cost = 0

        for i in range(len(route) - 1):
            cost += dist[(route[i], route[i+1])]

        if cost < min_cost:
            min_cost = cost
            min_path = route

    return min_path, min_cost


best_route, best_cost = tsp(1)

print("Best Route:", best_route)
print("Minimum Cost:", best_cost)
```

```python
#Task 3
graph = {
    1: [2, 3, 4],
    2: [1, 4, 3],
    3: [1, 4, 2],
    4: [1, 2, 3]
}


def dls(graph, node, goal, depth, path):
    if depth == 0 and node == goal:
        return path + [node]
    if depth > 0:
        for neighbor in graph.get(node, []):
            result = dls(graph, neighbor, goal, depth - 1, path + [node])
            if result:
                return result
    return None


def iddfs(graph, start, goal, max_depth):
    for depth in range(max_depth + 1):
        result = dls(graph, start, goal, depth, [])
        if result:
            return result
    return None


print("IDDFS Path (Graph):", iddfs(graph, 1, 3, 3))
```
