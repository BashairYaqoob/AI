```python
# AI Lab 3:
# 24k-0810
# Task 1
import random as r
components=['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I']

class Environment:
    def __init__(self):
        self.system={}
        for c in components:
            self.system[c]=r.choice(["Safe", "Vulnerable"])
    def get_percept(self):
        return self.system
class Agent:
    def act(self, percept):
        vulnerable=[]
        print("Running system scan:")
        for c, status in percept.items():
            if status=="Vulnerable":
                print(c, " is vulnerable")
                vulnerable.append(c)
            else:
                print(c, " is safe. Scan successfull!")   
        print("\nPatching Vulnerablities")
        for c in vulnerable:
            percept[c] = "Safe"
            print(c, " is patched")    

def run_agent(agent, environment):
    print("Initial State:", environment.system)
    percept = environment.get_percept()
    agent.act(percept)
    print("Final State:", environment.system)

e1 = Environment()
a1 = Agent()
run_agent(a1, e1)
```

```python
#Task 2
import random as r
class Environment:
    def __init__(self):
        self.servers={}
        states=["Underload", "Overload", "Balanced"]
        for i in range(5):
            self.servers[f"Server {i+1}"]=r.choice(states)
    def get_percept(self):
        return self.servers
class Agent:
    def act(self, percept):
        under=[]
        over=[]
        for s, state in percept.items():
            if state=="Underload":
                under.append(s)
            elif state=="Overload":
                over.append(s)
        for o in over:
            if under:
                u=under.pop(0)
                percept[o]="Balanced"
                percept[u]="Balanced"

def run_agent(environment, agent):
    print("Before Balancing:", environment.servers)
    agent.act(environment.get_percept())
    print("After Balancing:", environment.servers)

env1= Environment()
a1= Agent()
run_agent(env1, a1)
```

```python
#Task 3
class Environment:
    def __init__(self):
        self.tasks=["Completed", "Failed", "Completed", "Failed"]
    def get_percept(self):
        return self.tasks

class Agent:
    def act(self, percept):
        for i in range(len(percept)):
            if percept[i]=="Failed":
                print("Retrying backup ", i)
                percept[i] = "Completed"
def run_agent(environment, agent):
    print("Before:", environment.tasks)
    agent.act(environment.get_percept())
    print("After:", environment.tasks)
run_agent(Environment(), Agent())
```

```python
#Task 4
import random as r
components=['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I']

class Environment:
    def __init__(self):
        self.system={}
        for c in components:
            self.system[c]=r.choice(["Safe", "Low Risk Vulnerable", "High Risk Vulnerable"])
    def get_percept(self):
        return self.system
class Agent:
    def act(self, percept):
        print("Running system scan:")
        for c, status in percept.items():
            if status=="Low Risk Vulnerable":
                percept[c]="Safe"
                print(c, " patched")
            elif status== "High Risk Vulnerable":
                print(c, " is High Risk Vulnerable. Premium service needed!")
            else:
                print(c, " is safe. Scan successfull!")     

def run_agent(agent, environment):
    print("Initial State:", environment.system)
    percept = environment.get_percept()
    agent.act(percept)
    print("Final State:", environment.system)

e1 = Environment()
a1 = Agent()
run_agent(a1, e1)
```

```python
#Task 5
class Environment:
    def __init__(self):
        self.schedule = {
            "Room101": {"medicine": "Paracetamol", "time": "9 AM", "staff available":True},
            "Room102": {"medicine": "Insulin","time": "9 PM", "staff available":False}
            }
    def get_percept(self):
        return self.schedule
class Agent:
    def act(self, percept):
        for room, info in percept.items():
            medicine=info["medicine"]
            time=info["time"]
            staff=info["staff available"]
            print("Task received:")
            print("Room:", room)
            print("Medicine:", medicine)
            print("Time:", time)
            print("Moving to loaction")
            print("Picking", medicine)
            print("Moving to", room)
            print("Scanning patient ID")
            print("Delivered", medicine)
            if not staff:
                print("Staff not available! Alerting nurses")
            print()
def run_agent(agent, environment):
    agent.act(environment.get_percept())

run_agent(Agent(), Environment())
```

```python
#Task 6
class Environment:
    def __init__(self):
        self.rooms = {
            'a': ' ', 'b': ' ', 'c': '🔥',
            'd': ' ', 'e': '🔥', 'f': ' ',
            'g': ' ', 'h': ' ', 'j': '🔥'
        }
        self.path = ['a','b','c','d','e','f','g','h','j']

    def get_percept(self):
        return self.rooms

class Agent:
    def act(self, percept, path):
        for room in path:
            print("Robot enters", room)
            if percept[room] == "🔥":
                print("Fire detected and extinguished")
                percept[room] = " "


def run_agent(agent, environment):
    print("Initial:", environment.rooms)
    agent.act(environment.get_percept(), environment.path)
    print("Final:", environment.rooms)


run_agent(Agent(), Environment())
```
